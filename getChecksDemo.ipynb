This is a demo file for finding getting the checks for a circuit. It returns the list of checks. Works for Qiskit 0.23.3.
This demo creates a random Clifford circuit and then finds the desired number of gates for it.

```python
import qiskit
import checksfinder as chcksf
from qiskit.quantum_info import random_clifford, decompose_clifford
from qiskit.quantum_info.operators.symplectic.pauli_utils import pauli_basis
from multiprocessing import Pool
import psutil
```

```python
# Parameters
numQubits=5
maxPairs=3
circ=decompose_clifford(random_clifford(numQubits)) # Circ must be in the Clifford + T basis.
PARALLEL=False
print(circ)
```

There are two scenarios: parallel and not parallel. Parallel is probably overkill for searches for 5 qubits or less.

```python
pauli_table=pauli_basis(numQubits)
pauli_labels=pauli_table.to_labels() # convert to strings

checks_finder=chcksf.ChecksFinder(numQubits, circ)
if PARALLEL:
    pool=Pool(psutil.cpu_count(logical=False))
    print('starting parallel search...')
    count, p2_weights, pauli_str_p1s, pauli_str_p2s=checks_finder.get_min_checks_parallel(pool, pauli_labels, maxPairs)
else:
    count, p2_weights, pauli_str_p1s, pauli_str_p2s=checks_finder.get_min_checks_linear(pauli_labels, maxPairs)

print(f"left checks: {pauli_str_p1s}")
print(f"right checks: {pauli_str_p2s}")
print(f"count: {count}")
print(f"p2 weights: {p2_weights}")
```
